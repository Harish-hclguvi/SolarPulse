```python
def calculate_interval_coverage(df):

    df = df.sort_values(
        ["SOURCE_KEY_GEN", "DATE_TIME"]
    ).copy()

    grouped = df.groupby("SOURCE_KEY_GEN")

    df["TIME_GAP"] = grouped["DATE_TIME"].diff()

    coverage = df.groupby("SOURCE_KEY_GEN").agg(
        FIRST_READING=("DATE_TIME", "min"),
        LAST_READING=("DATE_TIME", "max")
    )

    valid_intervals = (
        df[df["TIME_GAP"] == pd.Timedelta(minutes=15)]
        .groupby("SOURCE_KEY_GEN")
        .size()
    )

    coverage["VALID_15MIN_INTERVALS"] = (
        valid_intervals.reindex(coverage.index, fill_value=0)
    )

    coverage["EXPECTED_INTERVALS"] = (
        (coverage["LAST_READING"] - coverage["FIRST_READING"])
        / pd.Timedelta(minutes=15)
    ).round().astype(int)

    coverage["COVERAGE_PERCENT"] = (
        coverage["VALID_15MIN_INTERVALS"]
        / coverage["EXPECTED_INTERVALS"]
        * 100
    )

    return coverage.reset_index()


plant1_coverage = calculate_interval_coverage(
    plant1_merged_clean
)

display(
    plant1_coverage.sort_values("COVERAGE_PERCENT")
)
```

```python
def calculate_solar_energy(df):

    df = df.sort_values(
        ["SOURCE_KEY_GEN", "DATE_TIME"]
    ).copy()

    # Previous AC power reading for each inverter
    df["PREVIOUS_AC_POWER"] = (
        df.groupby("SOURCE_KEY_GEN")["AC_POWER"].shift(1)
    )

    # Time difference between consecutive readings
    df["TIME_GAP"] = (
        df.groupby("SOURCE_KEY_GEN")["DATE_TIME"].diff()
    )

    # Identify valid 15-minute intervals
    df["IS_VALID_15MIN"] = (
        df["TIME_GAP"] == pd.Timedelta(minutes=15)
    )

    # Calculate energy only for valid intervals
    df["INTERVAL_ENERGY_KWH"] = np.where(
        df["IS_VALID_15MIN"],
        (
            (df["PREVIOUS_AC_POWER"] + df["AC_POWER"]) / 2
        ) * 0.25,
        np.nan
    )

    # Summarize energy by inverter
    energy_summary = (
        df.groupby("SOURCE_KEY_GEN")
        .agg(
            OBSERVED_ENERGY_KWH=(
                "INTERVAL_ENERGY_KWH", "sum"
            ),
            VALID_INTERVALS=(
                "IS_VALID_15MIN", "sum"
            )
        )
        .reset_index()
    )

    return df, energy_summary
```

### TOTAL_YIELD Data Quality Assessment

**Plant 1:**
- All 22 inverters have zero negative changes.
- The cumulative counter is non-decreasing.
- Retained for cautious trend analysis; units and absolute accuracy are not independently verified.

**Plant 2:**
- All 22 inverters have negative changes.
- Several inverters show extremely large positive and negative jumps.
- Some readings repeatedly fall to zero.
- Flagged as unreliable and excluded from cumulative-yield analysis.

**Final decision:**
- Retain TOTAL_YIELD in the datasets for traceability.
- Do not use Plant 2 TOTAL_YIELD for energy calculations or comparisons.
- Use AC_POWER-based interval energy as the primary energy metric.
- Do not delete or artificially correct anomalous readings.

```python
def calculate_hourly_plant_power(df, plant_name):

    # Aggregate inverter power at each timestamp
    timestamp_power = (
        df.groupby("DATE_TIME")
        .agg(
            TOTAL_PLANT_POWER=("AC_POWER", "sum"),
            INVERTER_COUNT=("SOURCE_KEY_GEN", "nunique")
        )
        .reset_index()
    )

    # Keep timestamps where all 22 inverters are available
    timestamp_power = timestamp_power[
        timestamp_power["INVERTER_COUNT"] == 22
    ].copy()

    # Extract hour
    timestamp_power["HOUR"] = (
        timestamp_power["DATE_TIME"].dt.hour
    )

    # Calculate average total plant power by hour
    hourly = (
        timestamp_power
        .groupby("HOUR")["TOTAL_PLANT_POWER"]
        .mean()
        .reset_index()
    )

    hourly["PLANT"] = plant_name

    return timestamp_power, hourly


plant1_timestamp_power, plant1_hourly_total = (
    calculate_hourly_plant_power(
        plant1_energy_data, "Plant 1"
    )
)

plant2_timestamp_power, plant2_hourly_total = (
    calculate_hourly_plant_power(
        plant2_energy_data, "Plant 2"
    )
)

hourly_plant_comparison = pd.concat(
    [plant1_hourly_total, plant2_hourly_total],
    ignore_index=True
)

display(hourly_plant_comparison)
```

```python
# Function to detect outliers using IQR
def detect_outliers_iqr(series):

    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = series[
        (series < lower_bound) |
        (series > upper_bound)
    ]

    return {
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "LOWER_BOUND": lower_bound,
        "UPPER_BOUND": upper_bound,
        "OUTLIER_COUNT": len(outliers),
        "OUTLIER_PERCENT": len(outliers) / len(series) * 100
    }


# Apply IQR method to both plants
plant1_outliers = detect_outliers_iqr(
    common_power["AC_POWER_PLANT1"]
)

plant2_outliers = detect_outliers_iqr(
    common_power["AC_POWER_PLANT2"]
)

outlier_comparison = pd.DataFrame({
    "Plant 1": plant1_outliers,
    "Plant 2": plant2_outliers
})

display(outlier_comparison.round(2))
```

```python
query15 = """
SELECT
    p1.DATE_TIME,
    p1.plant1_total_ac_power,
    p2.plant2_total_ac_power,
    ROUND(
        p1.plant1_total_ac_power - p2.plant2_total_ac_power,
        2
    ) AS power_difference
FROM (
    SELECT
        DATE_TIME,
        SUM(AC_POWER) AS plant1_total_ac_power
    FROM plant1_solar
    GROUP BY DATE_TIME
) AS p1
INNER JOIN (
    SELECT
        DATE_TIME,
        SUM(AC_POWER) AS plant2_total_ac_power
    FROM plant2_solar
    GROUP BY DATE_TIME
) AS p2
    ON p1.DATE_TIME = p2.DATE_TIME
ORDER BY p1.DATE_TIME
LIMIT 10;
"""

result15 = pd.read_sql(query15, engine)

display(result15)
```

```python
query16 = """
SELECT
    DATE(DATE_TIME) AS generation_date,
    ROUND(AVG(AC_POWER), 2) AS average_ac_power,
    ROUND(MAX(AC_POWER), 2) AS maximum_ac_power,
    ROUND(MIN(AC_POWER), 2) AS minimum_ac_power,
    COUNT(*) AS total_readings
FROM plant1_solar
GROUP BY DATE(DATE_TIME)
ORDER BY generation_date;
"""

result16 = pd.read_sql(query16, engine)

display(result16)
```
